## Research question and dataset
Given the same manually annotated ES regimes and causal market information, how do statistical, tree, neural, convolutional, recurrent, and attention models compare in predictive quality, temporal generalization, data efficiency, and computational cost?

ES.v.0 comes from Databento GLBX.MDP3. UTC one-minute OHLC is prepared into complete five-minute candles in the configured New York session. The intended policy requires trustworthy scheduled-status boundaries, an available dataset condition, a complete grid, and a valid prior-close reference chain. Historical fallback is retired. The current pipeline audit found 2,690 eligible sessions; final initialization must verify the completed annotated corpus.

Humans label broad Bull/Bear/Range regimes for both current and anticipated targets while viewing completed sessions. This is retrospective annotation; model inputs remain causal. No random candle splitting is allowed. Chronological development folds select configurations; all declared tracks freeze before the physically sealed final holdout can be explicitly opened.

Families: majority and prior-human-regime references, Gaussian Naive Bayes, logistic and polynomial logistic regression, kNN, decision tree, random forest, histogram gradient boosting, RBF SVM, MLP, TCN, GRU, and Transformer. All numerical results below come from saved artifacts.
